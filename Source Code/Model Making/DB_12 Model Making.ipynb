```python
from audiomentations import Compose, AddGaussianNoise, PitchShift

augment = Compose([
    AddGaussianNoise(min_amplitude=0.001, max_amplitude=0.01, p=0.5),
    PitchShift(min_semitones=-2, max_semitones=2, p=0.5),
])

def extract_embedding_augmented(file_path):
    waveform, sr = torchaudio.load(file_path)
    if sr != 16000:
        waveform = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)(waveform)
    audio = waveform.numpy()
    augmented = augment(samples=audio, sample_rate=16000)
    augmented_tensor = torch.tensor(augmented)
    if augmented_tensor.dim() == 1:
        augmented_tensor = augmented_tensor.unsqueeze(0)
    return ecapa.encode_batch(augmented_tensor).squeeze().detach().cpu().numpy()
```

```python
from pathlib import Path
from collections import defaultdict, Counter
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import numpy as np
import torchaudio

# ✅ Load ECAPA model once
from speechbrain.inference import SpeakerRecognition
ecapa = SpeakerRecognition.from_hparams(source="speechbrain/spkrec-ecapa-voxceleb", savedir="ecapa_model")

# ✅ Group files by speaker ID
def group_by_speaker(file_list, data_type="real"):
    grouped = defaultdict(list)
    for f in file_list:
        p = Path(f)
        if data_type == "real":
            speaker = p.parents[2].name
        elif data_type == "fake":
            speaker = p.parents[1].name
        grouped[speaker].append(f)
    return grouped

# ✅ Load + limit files for balance
real_files = load_audio_files("/content/drive/MyDrive/Project/dfdc_audio_extracted")
fake_all = timit_tts_files + deepfake_files

# ✅ Limit fakes to 10 per speaker to balance diversity
fake_speaker_map_full = group_by_speaker(fake_all, data_type="fake")
fake_files = []
for files in fake_speaker_map_full.values():
    fake_files.extend(files[:10])  # Limit

# ✅ Match real count to fake
real_files = real_files[:len(fake_files)]

# ✅ Rebuild speaker maps
real_speaker_map = group_by_speaker(real_files, data_type="real")
fake_speaker_map = group_by_speaker(fake_files, data_type="fake")

# ✅ Train/test speaker split
train_real, test_real = train_test_split(list(real_speaker_map), test_size=0.3, random_state=42)
train_fake, test_fake = train_test_split(list(fake_speaker_map), test_size=0.3, random_state=42)


```

```python
# ✅ Create labeled file pairs
train_files = [(f, 0) for spk in train_real for f in real_speaker_map[spk]] + \
              [(f, 1) for spk in train_fake for f in fake_speaker_map[spk]]
test_files = [(f, 0) for spk in test_real for f in real_speaker_map[spk]] + \
             [(f, 1) for spk in test_fake for f in fake_speaker_map[spk]]

print("Train distribution:", Counter([label for _, label in train_files]))
print("Test distribution:", Counter([label for _, label in test_files]))
```

```python
# ✅ High-quality embedding extractor (with resample & error handling)
def extract_embedding(file):
    try:
        waveform, sr = torchaudio.load(file)
        if sr != 16000:
            waveform = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)(waveform)
        if waveform.dim() == 1:
            waveform = waveform.unsqueeze(0)
        with torch.no_grad():
            emb = ecapa.encode_batch(waveform).squeeze().cpu().numpy()
        return emb
    except Exception as e:
        print(f"❌ Error extracting {file}: {e}")
        return None
```

```python
def extract_dataset(file_label_list, use_aug=False):
    X, y = [], []
    for f, label in tqdm(file_label_list, desc="🔍 Extracting embeddings"):
        try:
            if use_aug:
                emb = extract_embedding_augmented(f)
            else:
                emb = extract_embedding(f)
            X.append(emb)
            y.append(label)
        except Exception as e:
            print(f"⚠️ Skipping {f} due to error: {e}")
    return np.array(X), np.array(y)
```

```python
X_train, y_train = extract_dataset(train_files, use_aug=True)
X_test, y_test = extract_dataset(test_files)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier

# ✅ Normalize features (very important for MLP)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ✅ MLP with tuned depth + early stopping + dropout-style regularization
mlp_clf = MLPClassifier(
    hidden_layer_sizes=(512, 256, 64),  # deeper net
    activation='relu',
    alpha=0.001,              # L2 regularization
    learning_rate_init=0.001,
    max_iter=1000,
    random_state=42,
    early_stopping=True,
    validation_fraction=0.2,
    n_iter_no_change=20,
    verbose=True
)

mlp_clf.fit(X_train_scaled, y_train)
print("✅ MLPClassifier training complete.")
```

```python
from sklearn.metrics import (
    f1_score, accuracy_score, roc_auc_score,
    classification_report, roc_curve, confusion_matrix, ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
import numpy as np

# 🔮 Get predicted probabilities
probs = mlp_clf.predict_proba(X_test_scaled)[:, 1]

# 🎯 Tune threshold for best F1
best_f1, best_threshold = 0, 0
for t in np.arange(0.01, 1.0, 0.01):
    preds = (probs >= t).astype(int)
    f1 = f1_score(y_test, preds)
    if f1 > best_f1:
        best_f1 = f1
        best_threshold = t

# ✅ Final predictions
final_preds = (probs >= best_threshold).astype(int)

# ✅ Evaluation
acc = accuracy_score(y_test, final_preds)
f1 = f1_score(y_test, final_preds)
auc = roc_auc_score(y_test, probs)

print(f"\n🎯 Best Threshold = {best_threshold:.2f}")
print(f"✅ Accuracy: {acc:.4f} | F1: {f1:.4f} | AUC: {auc:.4f}")


# 📈 ROC Curve
fpr, tpr, _ = roc_curve(y_test, probs)
plt.figure(figsize=(6, 4))
plt.plot(fpr, tpr, label=f"AUC = {auc:.2f}")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.grid(True)
plt.legend()
plt.show()

```

```python
# 🧱 Confusion Matrix
cm = confusion_matrix(y_test, final_preds)
ConfusionMatrixDisplay(confusion_matrix=cm).plot()
plt.title("Confusion Matrix")
plt.grid(False)
plt.show()


# 📊 Classification Report
print("\n📃 Classification Report:\n")
print(classification_report(y_test, final_preds, target_names=["Real", "Fake"]))

```

```python
import torchaudio

def detect_audio(file_path, model=mlp_clf, threshold=0.5):
    # Load audio
    waveform, sr = torchaudio.load(file_path)

    # Resample to 16kHz if needed
    if sr != 16000:
        waveform = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)(waveform)

    # Ensure shape [1, N]
    if waveform.dim() == 1:
        waveform = waveform.unsqueeze(0)

    # Extract ECAPA embedding
    emb = ecapa.encode_batch(waveform).squeeze().detach().cpu().numpy()

    # Scale using training scaler
    scaled_emb = scaler.transform([emb])

    # Predict probability
    prob = model.predict_proba(scaled_emb)[0][1]

    # Decision
    label = "Fake" if prob >= threshold else "Real"
    print(f"\n🎧 File: {file_path}")
    print(f"🎯 Probability of being Fake: {prob:.4f}")
    print(f"✅ Prediction: {label} (Threshold: {threshold})")

    return label, prob
```

```python
def detect_audio(file_path, model, scaler, ecapa_model):
    """
    Predict whether an audio file is real or fake using ECAPA-TDNN and MLP.
    """
    import torchaudio
    import numpy as np
    import torch

    # Load and resample
    waveform, sr = torchaudio.load(file_path)
    if sr != 16000:
        waveform = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)(waveform)
    if waveform.dim() == 1:
        waveform = waveform.unsqueeze(0)

    # Extract ECAPA embedding
    with torch.no_grad():
        embedding = ecapa_model.encode_batch(waveform).squeeze().detach().cpu().numpy()

    # Normalize embedding
    embedding_scaled = scaler.transform([embedding])

    # Predict probability
    prob = model.predict_proba(embedding_scaled)[0][1]
    label = "Fake" if prob >= 0.5 else "Real"

    print(f"🔍 Prediction: {label} (Confidence: {prob:.4f})")
    return label, prob
```
